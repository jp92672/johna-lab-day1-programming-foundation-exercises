```python
# Modules for generating random values and timestamps
import random
import datetime

# Defines a device (sensor) and state
class Sensor:
    # Defines the sensor type, location, and value
    def __init__(self, sensor_type, location):
        self.sensor_type = sensor_type
        self.location = location
        self.value = None

    # Define a value or generates a random one
    def read(self, value=None):
        if value is not None:
            self.value = value
        elif self.sensor_type == "temperature":
            self.value = random.uniform(-5, 45)
        elif self.sensor_type == "motion":
            self.value = random.choice(["No activity", "Motion detected"])
        elif self.sensor_type == "smoke detector":
            self.value = random.choice(["No activity", "Smoke detected"])
        elif self.sensor_type == "door":
            self.value = random.choice(["Closed", "Door open"])
        elif self.sensor_type == "window":
            self.value = random.choice(["Closed", "Window open"])
        elif self.sensor_type == "light":
            self.value = random.choice(["On", "Off"])
        return self.value   

    # Checks if current value is abnormal
    def isAbnormal(self):
        if self.sensor_type == "temperature":
            return self.value < 0 or self.value > 40
        elif self.sensor_type == "motion":
            return self.value == "Motion detected"
        elif self.sensor_type == "smoke detector":
            return self.value == "Smoke detected"
        elif self.sensor_type == "door":
            return self.value == "Door open"
        elif self.sensor_type == "window":
            return self.value == "Window open"
        elif self.sensor_type == "light":
            return self.value == "On"
        else:
            return False

# Defines a room and alert
class Room:
    # Defines the room name and mode
    def __init__(self, name, mode):
        self.name = name
        self.mode = mode
        self.sensors = []
    
    # Loops through sensors and prints an alert
    def alerts(self):
        for sensor in self.sensors:
            if sensor.sensor_type == "temperature" and sensor.isAbnormal():
                print(f"⚠️ ALERT: Temperature abnormal in {self.name}!")
            elif sensor.sensor_type == "temperature" and self.mode == "home" and (sensor.value < 18 or sensor.value > 24):
                print(f"ℹ️ NOTICE: Temperature outside comfort zone in {self.name}!")
            elif sensor.sensor_type == "motion" and sensor.isAbnormal() and self.mode == "away":
                print(f"⚠️ ALERT: Motion detected in {self.name} while in away mode!")
            elif sensor.sensor_type == "smoke detector" and sensor.isAbnormal():
                print(f"‼️ ALARM: Smoke detected in {self.name}!")
            elif sensor.sensor_type == "door" and sensor.isAbnormal() and self.mode == "away":
                print(f"⚠️ ALERT: Door open in {self.name} while in away mode!")
            elif sensor.sensor_type == "window" and sensor.isAbnormal() and self.mode == "away":
                print(f"⚠️ ALERT: Window open in {self.name} while in away mode!")
            elif sensor.sensor_type == "light" and sensor.isAbnormal() and self.mode == "away":
                print(f"⚠️ ALERT: Light is on in {self.name} while in away mode!")
```

```python
# Room 1 set in away mode with all sensor types
room1_room = Room("Living Room", "away")
room1_room.sensors.append(Sensor("temperature", "Living Room"))
room1_room.sensors.append(Sensor("motion", "Living Room"))
room1_room.sensors.append(Sensor("smoke detector", "Living Room"))
room1_room.sensors.append(Sensor("door", "Living Room"))
room1_room.sensors.append(Sensor("window", "Living Room"))
room1_room.sensors.append(Sensor("light", "Living Room"))

# Room 2 set in home mode with all sensor types
room2_room = Room("Bedroom", "home")
room2_room.sensors.append(Sensor("temperature", "Bedroom"))
room2_room.sensors.append(Sensor("motion", "Bedroom"))
room2_room.sensors.append(Sensor("smoke detector", "Bedroom"))
room2_room.sensors.append(Sensor("door", "Bedroom"))
room2_room.sensors.append(Sensor("window", "Bedroom"))
room2_room.sensors.append(Sensor("light", "Bedroom"))

# Run 3 times to simulate variable readings and alerts, for every room
all_rooms = [room1_room, room2_room]
for i in range(3):
    print("=== HomeGuard Security System ===")
    print(f"Time: {datetime.datetime.now().strftime('%H:%M:%S')}")
    print()
    for room in all_rooms:
        print(f"Zone: {room.name} | Mode: {room.mode.upper()}")
        for sensor in room.sensors:
            sensor.read()
            print(f"[READING] {sensor.location} {sensor.sensor_type}: {sensor.value}")
        room.alerts()
        print()
```
